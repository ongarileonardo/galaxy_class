```python
import re
import matplotlib.pyplot as plt

# --- Parse results.txt ---
def parse_results(path):
    data = {}
    with open(path, "r") as f:
        content = f.read()
    # Match lines like: key: [1.23, 4.56, ...]
    pattern = re.compile(r"(\w+):\s*\[(.*?)\]")
    for match in pattern.finditer(content):
        key, values = match.groups()
        data[key] = [float(v) for v in values.split(",")]
    return data

data = parse_results("src/results.txt")

# --- Figure 1: Losses ---
loss_keys = ["raw_train_loss", "raw_train_smooth_loss", \
              "aug_train_loss", "aug_train_smooth_loss"]

plt.figure(figsize=(8, 5))
for key in loss_keys:
    epochs = range(1, len(data[key]) + 1)
    plt.plot(epochs, data[key], marker="o", label=key)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss (raw, smooth, augmented)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("learning_results/train_losses.pdf")
plt.show()

val_loss_keys = ["raw_val_loss", "raw_val_smooth_loss", \
                 "aug_val_loss", "aug_val_smooth_loss"]
plt.figure(figsize=(8, 5))
for key in val_loss_keys:
    epochs = range(1, len(data[key]) + 1)
    plt.plot(epochs, data[key], marker="o", label=key)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Validation Loss (raw, smooth, augmented)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("learning_results/val_losses.pdf")
plt.show()

# --- Figure 2: Accuracies ---
acc_keys = ["raw_train_acc", "raw_train_smooth_acc", \
            "aug_train_acc", "aug_train_smooth_acc"]

plt.figure(figsize=(8, 5))
for key in acc_keys:
    epochs = range(1, len(data[key]) + 1)
    plt.plot(epochs, data[key], marker="o", label=key)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training Accuracy")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("learning_results/train_accuracies.pdf")
plt.show()

val_acc_keys = ["raw_val_acc", "raw_val_smooth_acc", \
                "aug_val_acc", "aug_val_smooth_acc"]
plt.figure(figsize=(8, 5))
for key in val_acc_keys:
    epochs = range(1, len(data[key]) + 1)
    plt.plot(epochs, data[key], marker="o", label=key)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Validation Accuracy")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("learning_results/val_accuracies.pdf")
plt.show()
```
